```python
import os
from langchain_community.document_loaders import PyPDFLoader, DirectoryLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_community.llms import HuggingFaceEndpoint
from langchain.chains import RetrievalQA
```

```python

# ── 1. CONFIG ───────────────────────────────────────────────
HF_TOKEN = "your_huggingface_token_here"
DOCS_FOLDER = "./docs"          # put your PDFs in this folder
CHROMA_FOLDER = "./chroma_db"   # vector DB will be saved here
```

```python
# ── 2. LOAD DOCUMENTS ───────────────────────────────────────
print("Loading documents...")
loader = DirectoryLoader(DOCS_FOLDER, glob="**/*.pdf", loader_cls=PyPDFLoader)
documents = loader.load()
print(f"Loaded {len(documents)} pages")
```

```python
# ── 3. SPLIT INTO CHUNKS ─────────────────────────────────────
print("Splitting into chunks...")
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_documents(documents)
print(f"Created {len(chunks)} chunks")
```

```python
# ── 4. EMBED AND STORE ───────────────────────────────────────
print("Embedding and storing in vector DB...")
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
vectordb = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    persist_directory=CHROMA_FOLDER
)
vectordb.persist()
print("Vector DB ready!")
```

```python
# ── 5. SET UP LLM ────────────────────────────────────────────
print("Loading LLM...")
llm = HuggingFaceEndpoint(
    repo_id="mistralai/Mistral-7B-Instruct-v0.2",
    huggingfacehub_api_token=HF_TOKEN,
    temperature=0.5,
    max_new_tokens=512
)
```

```python
# ── 6. BUILD RAG CHAIN ───────────────────────────────────────
qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=vectordb.as_retriever(search_kwargs={"k": 4}),
    return_source_documents=True
)
```

```python
# ── 7. ASK QUESTIONS ─────────────────────────────────────────
print("\nRAG system ready! Type 'quit' to exit.\n")
while True:
    question = input("Your question: ")
    if question.lower() == "quit":
        break

    result = qa_chain.invoke(question)

    print("\nAnswer:", result["result"])
    print("\nSources:")
    for doc in result["source_documents"]:
        source = doc.metadata.get("source", "unknown")
        page = doc.metadata.get("page", "?")
        print(f"  - {source}, page {page}")
    print("\n" + "-"*50 + "\n")
```
